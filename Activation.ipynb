```python

import numpy as np

# Input values
inputs = np.array([2, 3])

# ReLU activation function
def relu(x):
    return np.maximum(0, x)

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Hidden layer weights and biases
weights_hidden = np.array([
    [0.5, 0.4],
    [0.3, -0.5]
])

bias_hidden = np.array([0.2, 0.1])

# Step 1: Calculate hidden layer weighted sums
z_hidden = np.dot(weights_hidden, inputs) + bias_hidden

# Step 2: Apply ReLU to hidden layer
a_hidden = relu(z_hidden)

print("Hidden layer weighted sums:", z_hidden)
print("Hidden layer outputs:", a_hidden)

# Output layer weights and bias
weights_output = np.array([0.6, 0.8])
bias_output = 0.1

# Step 3: Calculate output weighted sum
z_output = np.dot(weights_output, a_hidden) + bias_output

# Step 4: Apply sigmoid
final_output = sigmoid(z_output)

print("Output weighted sum:", z_output)
print("Final prediction:", final_output)
```
