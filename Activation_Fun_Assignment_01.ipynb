```python

import numpy as np

# Input values
inputs = np.array([2, 3])

# 1. ReLU activation function
def relu(x):
    return np.maximum(0, x)

# 2. Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# 3. Linear activation function
def linear(x):
    return x

# 4. Tanh activation function
def tanh(x):
    return np.tanh(x)


# Hidden layer weights and biases
weights_hidden = np.array([
    [0.5, 0.4],
    [0.3, -0.5]
])

bias_hidden = np.array([0.2, 0.1])

# Step 1: Calculate hidden layer weighted sums
z_hidden = np.dot(weights_hidden, inputs) + bias_hidden

print("Hidden layer weighted sums:", z_hidden)


# Step 2: Apply all four activation functions
a_relu = relu(z_hidden)
a_sigmoid = sigmoid(z_hidden)
a_linear = linear(z_hidden)
a_tanh = tanh(z_hidden)

print("\nReLU hidden layer output:", a_relu)
print("Sigmoid hidden layer output:", a_sigmoid)
print("Linear hidden layer output:", a_linear)
print("Tanh hidden layer output:", a_tanh)


# Output layer weights and bias
weights_output = np.array([0.6, 0.8])
bias_output = 0.1


# Step 3: Calculate output weighted sum
# Using ReLU activation
z_output_relu = np.dot(weights_output, a_relu) + bias_output

# Step 4: Apply sigmoid
final_output_relu = sigmoid(z_output_relu)

print("\nReLU Output weighted sum:", z_output_relu)
print("ReLU Final prediction:", final_output_relu)


# Using Sigmoid activation
z_output_sigmoid = np.dot(weights_output, a_sigmoid) + bias_output
final_output_sigmoid = sigmoid(z_output_sigmoid)

print("\nSigmoid Output weighted sum:", z_output_sigmoid)
print("Sigmoid Final prediction:", final_output_sigmoid)


# Using Linear activation
z_output_linear = np.dot(weights_output, a_linear) + bias_output
final_output_linear = sigmoid(z_output_linear)

print("\nLinear Output weighted sum:", z_output_linear)
print("Linear Final prediction:", final_output_linear)


# Using Tanh activation
z_output_tanh = np.dot(weights_output, a_tanh) + bias_output
final_output_tanh = sigmoid(z_output_tanh)

print("\nTanh Output weighted sum:", z_output_tanh)
print("Tanh Final prediction:", final_output_tanh)
```
