```python

import numpy as np

# Different input datasets
inputs_list = [
    np.array([2, 3]),
    np.array([-2, 1]),
    np.array([0, 4]),
    np.array([-3, -1])
]

# Activation functions
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def linear(x):
    return x

def tanh(x):
    return np.tanh(x)


# Hidden layer weights and biases
weights_hidden = np.array([
    [0.5, 0.4],
    [0.3, -0.5]
])

bias_hidden = np.array([0.2, 0.1])

# Output layer weights and bias
weights_output = np.array([0.6, 0.8])
bias_output = 0.1


# Store activation functions
activations = {
    "ReLU": relu,
    "Sigmoid": sigmoid,
    "Linear": linear,
    "Tanh": tanh
}


# Process each numerical dataset
for inputs in inputs_list:

    print("\n================================")
    print("Input values:", inputs)

    # Step 1: Hidden layer weighted sum
    z_hidden = np.dot(weights_hidden, inputs) + bias_hidden

    print("Hidden weighted sum:", z_hidden)

    # Step 2: Apply each activation function
    for name, activation in activations.items():

        a_hidden = activation(z_hidden)

        print("\n", name, "Hidden output:", a_hidden)

        # Step 3: Output layer weighted sum
        z_output = np.dot(weights_output, a_hidden) + bias_output

        # Step 4: Apply sigmoid to output
        final_output = sigmoid(z_output)

        print(name, "Output weighted sum:", z_output)   
        print(name, "Final prediction:", final_output)
```
