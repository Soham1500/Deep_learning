```python
import numpy as np

def sigmoid(x):
    return 1/(1+np.exp(-x))

def relu(x):
    return np.maximum(0,x)

weights_one=np.array([
    [2,3],
    [4,5]
])

x_value=np.array([2,3])

bias_one=np.array([1,2])

h_value=np.dot(weights_one,x_value)+bias_one

relu_value=relu(h_value)
print("relu_values = ",relu_value)

bias_two=2

weights_two=[2,8]

sigmoid_value=np.dot(weights_two,relu_value)+bias_two

print("sigmoid value ",(sigmoid(sigmoid_value)) )

```
